```python
# =========================
# STAGE 1 — PREPROCESSING
# =========================

!pip install groq langchain langchain_groq datasets python-dotenv

import json
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from google.colab import files

# ---------------------------------------------------------
# STEP 1 — Upload raw posts text file (one post per line)
# ---------------------------------------------------------
print("Upload your raw_posts.txt file:")
uploaded = files.upload()

raw_file = list(uploaded.keys())[0]

with open(raw_file, "r") as f:
    raw_posts = [line.strip() for line in f.readlines() if line.strip()]

raw_posts[:3]
```

```python
import os

os.environ["GROQ_API_KEY"] = "YOUR_GROQ_API_KEY"   # Replace manually

llm = ChatGroq(
    model="llama3-8b-8192",
    temperature=0.0
)
```

```python
prompt_template = """
Extract metadata for the LinkedIn post below:

Post:
{post}

Return ONLY valid JSON in the following format:
{{
  "post": "...",
  "topic": "",
  "language": "",
  "length": ""
}}
"""

prompt = PromptTemplate(
    input_variables=["post"],
    template=prompt_template
)
```

```python
enriched_posts = []

for post in raw_posts:
    chain = prompt | llm
    result = chain.invoke({"post": post})

    try:
        enriched = json.loads(result.content)
        enriched_posts.append(enriched)
    except:
        print("JSON Parse Error for post:", post)

len(enriched_posts)
```

```python
with open("enriched_posts.json", "w") as f:
    json.dump(enriched_posts, f, indent=4)

files.download("enriched_posts.json")
```

```python
!pip install streamlit

import streamlit as st
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv

load_dotenv()

st.title("📢 LinkedIn Post Generator - Llama 3.2 + Groq")

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    st.error("Missing GROQ_API_KEY in environment.")
    st.stop()

llm = ChatGroq(
    model="llama3-8b-8192",
    temperature=0.7
)

# ------------------------------
# UI Controls
# ------------------------------
topic = st.text_input("Topic")
tone = st.selectbox("Tone", ["Professional", "Friendly", "Funny", "Inspirational"])
length = st.selectbox("Length", ["Short", "Medium", "Long"])

generate_btn = st.button("Generate LinkedIn Post")

# ------------------------------
# Template for generation
# ------------------------------
template = """
Create a LinkedIn post based on the following:

Topic: {topic}
Tone: {tone}
Length: {length}

Your response MUST be a natural, well-written LinkedIn post.
"""

prompt = PromptTemplate(
    input_variables=["topic", "tone", "length"],
    template=template
)

# ------------------------------
# GENERATE POST
# ------------------------------
if generate_btn:
    chain = prompt | llm
    response = chain.invoke({
        "topic": topic,
        "tone": tone,
        "length": length
    })

    st.subheader("Generated Post:")
    st.write(response.content)
```
