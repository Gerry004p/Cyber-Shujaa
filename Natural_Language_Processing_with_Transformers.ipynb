```python
# Install spaCy and download the English language model
# !pip install -u spacy
#!python -m spacy download en_core_web_sm

#import necessary libraries
import spacy
from spacy import displacy
from transformers import BertTokenizer, TFBertModel
import tensorflow as tf
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Load pre-trained BERT tokenizer and BERT model from Hugging Face
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased', force_download=True)
bert_model = TFBertModel.from_pretrained('bert-base-uncased', force_download=True, use_safetensors=False)
```

```python
# Example sentence pairs (some similar, some not) added 5 more of my own pairs
sentence_pairs = [
    ("How do I learn Python?", "What is the best way to study Python?"),
    ("What is AI?", "How to cook pasta?"),
    ("How do I bake a chocolate cake?", "Give me a chocolate cake recipe."),
    ("How can I improve my coding skills?", "Tips for becoming better at programming."),
    ("Where can I buy cheap laptops?", "Best sites to find affordable computers."),
    ("What is machine learning?","Tips for cooking better."),
    ("Who is a Statician?","Techniques to understand calculus."),
    ("Is education relevant?","Benefits of having that degree."),
    ("How to cook rice and stew?","What is deep learning?"),
    ("What are the importance of novels?","Who is a chef?")
]

# Ground truth similarity labels: 1 = similar, 0 = not similar
labels = [1, 0, 1, 1, 1,0,1,1,0,0]
```

```python
# Function to get the BERT [CLS] embedding for a sentence
def get_sentence_embedding(sentence):
    # Tokenize and encode sentence into input tensors
    inputs = tokenizer(sentence, return_tensors='tf', padding=True, truncation=True)
    # Get model output
    outputs = bert_model(inputs)
    # Extract [CLS] token embedding (shape: [1, 768])
    cls_embedding = outputs.last_hidden_state[:, 0, :]
    return cls_embedding.numpy()
```

```python
# Calculate cosine similarity for each pair
predictions = []
for sent1, sent2 in sentence_pairs:
    emb1 = get_sentence_embedding(sent1)
    emb2 = get_sentence_embedding(sent2)
    sim_score = cosine_similarity(emb1, emb2)[0][0]
    pred = 1 if sim_score > 0.7 else 0
    predictions.append(pred)

    print(f"\nSentence 1: {sent1}")
    print(f"Sentence 2: {sent2}")
    print(f"Cosine Similarity: {sim_score:.4f} → Predicted Similar: {pred}")
```

```python
how does BERT differ from traditional NLP approaches like bag of words or TF-IDF What is the role of the encoder in the BERT model and how it was use in the above assignment What are the contextual embeddings and how are they generated ad used in the code Why is the [CLS] token used for sentence similarity in the code What is the cosine similarity and why is it useful i comparing embeddings

How does BERT differ from traditional NLP approaches like Bag of Words or TF-IDF?
BERT (Bidirectional Encoder Representations from Transformers) differs significantly from traditional NLP approaches like Bag of Words (BoW) and TF-IDF in several key aspects:

Contextual Understanding:

BoW/TF-IDF: These methods treat words in isolation. "Bank" in "river bank" would have the same representation as "bank" in "money bank." They completely lose the word order and context.
BERT: BERT is deeply contextual. It processes words bidirectionally, considering the words that come before and after a given word in a sentence. This means "bank" gets a different numerical representation depending on whether it's related to a river or a financial institution. This contextual understanding is crucial for nuanced language comprehension.
Word Embeddings:

BoW/TF-IDF: These generate sparse vector representations (often high-dimensional with many zeros) that simply count word occurrences or their importance. They don't capture semantic relationships between words directly.
BERT: BERT generates dense, continuous vector embeddings (often 768 or 1024 dimensions) for each word, which are dynamically generated based on the word's context. These embeddings capture rich semantic and syntactic information, meaning words with similar meanings or grammatical roles will have similar embeddings.
Architecture:

BoW/TF-IDF: These are feature engineering techniques, not deep learning models. They are relatively simple statistical methods.
BERT: BERT is based on the Transformer architecture, a deep neural network that uses attention mechanisms. This allows it to weigh the importance of different words in a sentence when processing another word, contributing to its strong contextual understanding.
Pre-training and Fine-tuning:

BoW/TF-IDF: These methods don't involve pre-training on massive datasets in the same way. They are typically applied directly to the specific dataset at hand.
BERT: BERT is pre-trained on a huge amount of text data (like Wikipedia and BookCorpus) using self-supervised tasks (e.g., Masked Language Modeling and Next Sentence Prediction). This pre-training allows it to learn general language understanding. After pre-training, it can be fine-tuned with a small, task-specific dataset for specific NLP tasks like sentiment analysis, question answering, or named entity recognition, often achieving state-of-the-art results with much less data than traditional methods would require.
Handling Out-of-Vocabulary (OOV) Words:

BoW/TF-IDF: These typically struggle with OOV words; new words not seen during training are usually ignored or assigned a default value.
BERT: BERT uses a subword tokenization strategy (e.g., WordPiece), breaking down words into smaller units. This allows it to handle OOV words more gracefully, as it can construct representations for them from known subword units.
What is the role of the encoder in the BERT model and how was it used in the above assignment?
BERT's architecture is primarily composed of multiple Transformer encoder layers stacked on top of each other. The role of the encoder is to process a sequence of input tokens and produce a sequence of contextualized representations (embeddings) for each token. Each encoder layer takes its input (either the initial embeddings or the output from the previous encoder layer) and applies self-attention mechanisms and feed-forward neural networks to transform these representations, effectively capturing relationships between all words in the input sequence.

In the provided assignment, the BERT encoder was used to:

Generate contextual embeddings for input sentences: The bert_model = TFBertModel.from_pretrained('bert-base-uncased', ...) line loads the pre-trained BERT encoder.
Extract sentence-level representations: The get_sentence_embedding function utilizes this encoder. When you pass a sentence through bert_model(inputs), the encoder processes the tokenized input and returns outputs.last_hidden_state. This last_hidden_state contains the contextualized embedding for each token in the input sentence, where each embedding reflects the meaning of the token in the context of the entire sentence.
What are contextual embeddings and how are they generated and used in the code?
Contextual embeddings are vector representations of words (or subwords) where the vector for a word changes depending on the other words surrounding it in a given sentence. Unlike traditional static embeddings (like Word2Vec or GloVe) where a word always has the same vector regardless of context, contextual embeddings capture polysemy (multiple meanings) and semantic nuances.

How they are generated in BERT: BERT generates contextual embeddings through its multi-layered Transformer encoder. Each token in the input sequence is initially represented by a combination of its token embedding, segment embedding (indicating if it's from sentence A or B), and positional embedding (indicating its position in the sentence). These combined embeddings then pass through multiple self-attention layers within the encoder. The self-attention mechanism allows each token's representation to be influenced by every other token in the input sequence, resulting in a vector that is highly dependent on its specific context.

How they are used in the code: In the get_sentence_embedding function, after the input sentence is tokenized and passed through the bert_model (which is the BERT encoder), outputs.last_hidden_state contains the contextual embeddings for all tokens. Specifically, the code then extracts the embedding of the [CLS] token (the first token in the sequence) as cls_embedding = outputs.last_hidden_state[:, 0, :]. This [CLS] token's embedding is considered to be a good aggregate representation of the entire sentence's meaning, as it has attended to all other tokens during the encoder's processing.

Why is the [CLS] token used for sentence similarity in the code?
The [CLS] (classifier) token is a special token prepended to the beginning of every input sequence to BERT. During BERT's pre-training, particularly for the Next Sentence Prediction (NSP) task, the final hidden state corresponding to this [CLS] token is used as the aggregate representation of the entire input sequence for classification tasks. Because it's trained to summarize the entire input sequence for tasks like NSP, its final embedding is often considered a semantically rich representation of the entire sentence.

Therefore, in the provided code, the embedding of the [CLS] token (outputs.last_hidden_state[:, 0, :]) is extracted to serve as the sentence embedding for each input sentence. These sentence embeddings are then used to calculate similarity between sentences.

What is cosine similarity and why is it useful in comparing embeddings?
Cosine similarity is a measure of similarity between two non-zero vectors of an inner product space. It is calculated as the cosine of the angle between two vectors. The value ranges from -1 to 1:

1: Indicates that the vectors are identical in direction (maximum similarity).
0: Indicates that the vectors are orthogonal (no similarity).
-1: Indicates that the vectors are diametrically opposed (maximum dissimilarity).
Why it's useful for comparing embeddings:

Direction over Magnitude: When comparing embeddings (like those generated by BERT), the direction of the vectors often signifies semantic meaning more than their magnitude. Words or sentences with similar meanings tend to have embeddings that point in similar directions in the high-dimensional space. Cosine similarity focuses purely on the angle between vectors, making it robust to differences in vector length.
Semantic Similarity: It effectively quantifies how semantically similar two words, phrases, or sentences are. If two embeddings are close in meaning, their cosine similarity will be high (close to 1).
Dimensionality Reduction Implicitly: For high-dimensional vectors, Euclidean distance can be misleading. Cosine similarity provides a more reliable measure of conceptual closeness in such spaces.
In the code, cosine_similarity(emb1, emb2)[0][0] calculates this measure between the [CLS] embeddings of two sentences. A high cosine similarity score implies that the two sentences are semantically similar.

```
